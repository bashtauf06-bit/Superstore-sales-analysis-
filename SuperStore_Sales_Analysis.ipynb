```python
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('ggplot')
```

```python
df = pd.read_csv('Sample-Superstore.xls.csv')
```

```python
print("Dataset loaded! Shape:", df.shape)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
```

```python
df = df.dropna()
```

```python
print("Cleaned dataset shape:", df.shape)
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(5)
```

```python
print("\nTop 5 Products by Sales:")
```

```python
print(top_products)
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales.plot(kind='line', title='Monthly Sales Trend', figsize=(10,5))
plt.ylabel('Sales ($)')
plt.show()
```

```python
print(df.columns)
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(5)
print("\nTop 5 Products by Sales:")
print(top_products)
```

```python
sales_region = df.groupby('Region')[['Sales']].sum().sort_values(by='Sales')
sales_region.plot(kind='barh', title='Total Sales by Region', figsize=(8,4))
plt.xlabel('Sales ($)')
plt.show()
```
